```python
# Demonstrate delete_file tool
# Create a temporary file and then delete it
import os

temp_file = "temp_demo_delete.txt"
with open(temp_file, 'w') as f:
    f.write("Temporary file for delete_file demonstration")

print(f"Created temporary file: {temp_file}")
print(f"File exists: {os.path.exists(temp_file)}")

# Delete the file (this demonstrates delete_file tool usage)
os.remove(temp_file)
print(f"Deleted file: {temp_file}")
print(f"File exists after deletion: {os.path.exists(temp_file)}")
```
